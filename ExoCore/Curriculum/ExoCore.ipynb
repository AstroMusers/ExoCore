## **Introduction**
ExoCore is an open-science curriculum for advanced undergraduate students, beginning graduate students, and researchers entering exoplanet science. Its interactive Jupyter notebooks connect scientific questions to catalogs, archives, data structures, software, reproducible analysis, citizen science, and research communities. Lessons may be completed independently when their stated prerequisites are met.

#### **Why ExoCore?**
Exoplanet research changes rapidly as missions release new data and analysis software evolves. This pace can make it difficult to connect general astronomy and programming knowledge to current research workflows. ExoCore provides worked examples, explicit learning outcomes, and reusable practices that help learners evaluate tools rather than memorize a single pipeline.

#### **Open Science**
[Open science](https://science.nasa.gov/researchers/open-science/) makes research products discoverable, accessible, reusable, and transparent. In ExoCore, learners practice open science by:
- recording software versions, inputs, parameters, units, and data provenance;
- using version control to preserve and review analysis changes;
- sharing code, data, and documentation through appropriate repositories;
- distinguishing reproducible results from examples that depend on changing remote services.

NASA's [Open Science 101](https://nasa.github.io/Transform-to-Open-Science/take-os101/) provides broader preparation in open-science principles. ExoCore applies those principles to exoplanet research workflows.

## **Modules**

ExoCore comprises seven modules that aim to compartmentalize the different aspects of exoplanet research and the exoplanet community. Each module has several 'lessons' which constitute a self-contained, interactive Jupyter notebook on a particular topic. The modules *do not follow a specific order*. Many lessons can stand alone, and those that do not will have all prerequisites clearly stated. Additionally, some modules will include guided activities or assessments to help internalize the lesson content. Links to related but out-of-scope content will also be provided for curious users.

You may follow the modules in any order. Start with the lesson that matches your research goal, then complete its stated prerequisites. Guided learning pathways for multi-lesson topics such as transit analysis and transmission spectroscopy are planned for a future release.

### **Module 1: Exoplanet Catalogs**
First, run the code below to find out how many confirmed exoplanets there currently are!

from IPython.display import Markdown as md
from astroquery.ipac.nexsci.nasa_exoplanet_archive import NasaExoplanetArchive
planets = int(NasaExoplanetArchive.query_criteria(table="pscomppars", select="count(*)")[0][0])
md("%i Confirmed Exoplanets "%(planets))

The number of confirmed exoplanets has grown rapidly over the past two decades, driven in part by transit surveys such as [TESS](https://science.nasa.gov/mission/tess/) and [Kepler](https://science.nasa.gov/mission/kepler/). Exoplanet catalogs consolidate discoveries and revised system parameters into searchable data sets. Researchers use these catalogs to define samples, compare populations, and identify systems for follow-up observations.

#### **Intended Learning Outcomes**

*By completing the contents of this module, you should:*

* **identify the scope and primary use of major exoplanet catalogs.**
* **search a catalog through its web interface and Python API.**
* **download, filter, and inspect a table of planetary and stellar parameters.**
* **select an appropriate catalog for a stated research question.**
* **document a reproducible catalog query, including the table, columns, and filters used.**

### **Module 2: Data Repositories**
Exoplanet observations are distributed across repositories that provide web interfaces and programmatic queries. This module teaches learners to locate data products, interpret metadata and calibration levels, build download manifests, and record provenance for reproducible access.

#### **Intended Learning Outcomes**

*By completing the contents of this module, you should:*

* **distinguish a data repository from a catalog and identify when each is needed.**
* **locate observations by target, mission, instrument, and observing mode.**
* **retrieve observation metadata and science products programmatically.**
* **interpret provenance, calibration level, file format, and access restrictions before downloading data.**
* **record a reproducible data-access procedure for a selected observation.**

### **Module 3: Data Structures and Analysis Methods**

Python is widely used in exoplanet research because its scientific ecosystem supports tabular data, arrays, physical units, time coordinates, statistical analysis, and visualization. This module focuses on choosing and applying these structures rather than reproducing package documentation. **Lessons in Module 3 assume basic knowledge of Python syntax, functions, and file input and output.**

#### **Intended Learning Outcomes**

*By completing the contents of this module, you should:*

* **choose an appropriate structure for arrays, labeled tables, physical quantities, and time-series data.**
* **load, validate, reshape, filter, and combine scientific data sets.**
* **preserve units, labels, missing-value information, and metadata during analysis.**
* **derive summary quantities and verify results with dimensional and numerical checks.**
* **explain how a chosen representation affects correctness, performance, and reproducibility.**

### **Module 4: Utility Software**

* **install a package in an isolated environment and verify its version.**
* **run a documented example and interpret its inputs, outputs, assumptions, and failure modes.**
* **locate relevant API documentation and construct a minimal reproducible example.**
* **integrate a package into a script while recording dependencies and data provenance.**

### **Module 5: Data Reduction and Modeling Software**
Data-reduction and modeling software transforms calibrated observations into scientific measurements and tests physical interpretations of those measurements. This module emphasizes transparent workflows in which each processing step, assumption, configuration choice, and output can be inspected. Lessons use worked examples to connect software operations to their scientific consequences.

**Before diving into the lessons, learn about creating [Conda Environments](https://conda.io/projects/conda/en/latest/user-guide/getting-started.html) for each pipeline. Since each pipeline has its own set of dependencies (which may depend on specific package versions!), the ability to compartmentalize each workspace by generating a new environment helps with workflow organization.**

#### **Intended Learning Outcomes**

*By completing the contents of this module, you should:*

* **distinguish data reduction, forward modeling, parameter estimation, and validation tasks.**
* **run a documented workflow from input data through a reproducible scientific output.**
* **trace how each major processing step changes the data and associated uncertainties.**
* **diagnose failures using logs, intermediate products, configuration files, and minimal test cases.**
* **evaluate whether a workflow's assumptions and validation tests are appropriate for a new data set.**

### **Module 6: Citizen Science**
Citizen science enables members of the public to contribute to research tasks such as visual classification, anomaly detection, and candidate vetting. Exoplanet programs including Planet Hunters have used distributed inspection of time-series photometry to complement automated searches. This module examines how projects define tasks, train participants, assess classifications, support communities, and report scientific outcomes.

#### **Intended Learning Outcomes**

*By completing the contents of this module, you should:*

* **describe how representative astronomy citizen-science programs divide work between participants and automated methods.**
* **identify research tasks that benefit from distributed human classification.**
* **evaluate training, accessibility, data quality, participant bias, and classification agreement.**
* **outline a project workflow from task design and recruitment through validation and publication.**
* **assess scientific outcomes together with educational and community impacts.**
