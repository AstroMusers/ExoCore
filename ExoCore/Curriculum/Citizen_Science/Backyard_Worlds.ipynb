A notable citizen science project on Zooniverse is [Backyard Worlds: Planet 9](https://www.zooniverse.org/projects/marckuchner/backyard-worlds-planet-9), a NASA-funded search for objects in the outer Solar System. Volunteers inspect images from NASA's Wide-field Infrared Survey Explorer (WISE) and mark moving objects such as brown dwarfs, low-mass stars, and possible distant planets. These faint objects are difficult to detect. Many volunteers can help vet and identify candidates.

Volunteers mark moving objects and dipoles in the images. The tutorial explains what to mark and what to leave alone. Short, clear tasks help participants make consistent classifications.

Backyard Worlds looks for objects in image sequences, while Planet Hunters projects search for transits in light curves. Image-processing software can miss faint movers or dipoles because of noise, and scattered light can create bright ghost images. Human reviewers can help distinguish these artifacts from candidate objects. Clyde Tombaugh discovered Pluto in 1930 by comparing photographic plates.
